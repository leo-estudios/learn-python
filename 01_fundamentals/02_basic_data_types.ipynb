```python
# Entero o int

"""
Los enteros son un tipo de dato que permite representar números enteros, es decir, positivos y negativos no decimales.
"""

i = 12
print(i) # -> 12
print(type(i)) # -> int

## Convertir a int

"""
Es posible convertir a int otro tipo. Como hemos explicado, el tipo int no puede contener decimales, por lo que intentamos convertir un 
número decimalm se truncará todo lo que tengamos a la derecha de la coma.
"""

b = int(1.6)
print(b) # -> 1
```

```python
# Booleanos

"""
Al igual que en otros lenguajes de programación, en Python existe el tipo bool o booleano. Es un tipo de dato que permite almacenar 
dos valores True o False.
"""

## Declarar la variable bool

x = True
y = False
```

```python
# Float

"""
El tipo numérico float permite representar un número positivo o negativo con decimales, es decir, números reales. Si vienes de otros 
lenguajes, tal vez conozcas el tipo doble, lo que significa que tiene el doble de precisión que un float. En Python las cosas son algo distintas, 
y los float son en realidad double.

Por lo tanto si declaramos una variable y le asignamos un valor decimal, por defecto la variable será de tipo float
"""

f = 0.10093
print(f) # -> 0.10093
print(type(f)) # -> float

## Convertir a float

"""
También se puede declarar usando la notación cientifica con e y el exponente. El siguente ejemplo sería lo mismo que decir 1.93 multiplicado 
por diez elevado a -3
"""

f = 1.93e-3

"""
También podemos convertir otro tipo a float haciendo uso de float(). Podemos ver como True es en realidad tratado como 1, y al convertirlo a 
float, como 1.0.
"""

a = float(True)
b = float(1)

print(a, type(a)) # -> 1.0, float
print(b, type(b)) # -> 1.0, float

## Rango respetable

"""
Alguna curiosidad es que los float no tienen precisión infinita. Podemos ver en el ejemplo como en realidad f se almacena como si fuese 1, ya 
que no es posible representar tanta precisión decimal.
"""

f = 0.9999999999999999999
print(f) # -> 1.0
print(1 == f) # -> True
```

```python
# Strings

"""
Son un tipo inmutable que permite almacenar secuencias de caracteres. Para crear una, es necesario incluir el texto entre comillas dobles.
"""

s = "Esto es una cadena"
print(s) # -> Esto es una cadena
print(type(s)) # -> str

"""
También es valido declararlo entre cadenas simples.

Las cadenas no están limitadas, por lo que el único límite es la memoria de tu ordenador. Una cadena puede estar vacia también.
"""

s = 'Esto es otra cadena'
print(s) # -> Esto es otra cadena
print(type(s)) # -> str

empty_s = ''
print(empty_s) # -> 
print(type(empty_s)) # -> str
```
