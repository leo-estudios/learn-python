```python
# Variables en Python

"""
Las variables se utilizan para almacenar datos que pueden consultarse y manipularse durante la ejecución del un programa. Una 
variable es, esencialmente, un nombre asignado a un valor.

· A diferencia de Java y muchos otros lenguajes, las variables de Python no requieren una declaración explícita de su tipo
· El tipo de variable se deduce en función del valor asignado
"""

x = 5
name = "Alex"

print(x) # 5
print(name) # Alex
```

```python
# Reglas para nombrar variables

"""
Para utilizar las variables correctamente, se deben seguir las siguientes reglas de nomenclatura:

1. Los nombres pueden contener letras, dígitos y guinoes bajos.
2. El primer carácter no puede ser un dígito.
3. Los nombres distinguen entre mayúsuculas y minúsculas, por lo que "myVar" y "myvar" se tratan de forma diferente.
4. Las palabras clave como if, else y for no se pueden utilizar como nombres de variables.
"""

## Los nombres de las variables que se enumeran a continuación son válidos:
age = 21
_colour = "lilac"
total_score = 90

## Los nombres de las variables que se enumeran a continuación no son válidos:
# 1name = "Error" # -> comienza con un dígito
# class = 10 # -> la palabra class es una palabra reservada
# user-name = "Doe" # -> contiene guíones
```

```python
# Asignación de valores a variables

## 1. Asignación básica: a las variables se les asignan valores usando el operador =
x = 5
y = 3.14
z = "Hi"

## 2. Tipado dinámico: nos ofrece la posibilidad de que una misma variable pueda almacenar diferentes tipos de datos durante la ejecución
x = 10
x = "Now a string"

## 3. Asignación del mismo valor: se puede asignar el mismo valor a varias variables en una sola línea
a = b = c = 100
print(a, b, c)

## 4. Asignación de valores diferentes: también se pueden asignar valores diferentes a varias variables en una sola línea
x, y, z = 1, 2.5, "Python"
print(x, y, z)
```
