```python
# Strings

"""
Las cadenas o strings son un tipo inmutable que permite almacenar secuencias de caracteres. Para crear una, 
es necesario incluir el texto entre comilas dobles ". Se puede obtener información sobre este tipo usando 
el comando help(str).
"""

s = "Esto es una cadena"
print(s) # -> Esto es una cadena
print(type(s)) # -> str

"""
También es valido declarar las cadenas con comillas simples '.
"""

s = 'Esto es otra cadena de texto'
print(s) # -> Esto es otra cadena de texto
print(type(s)) # -> str

"""
Las cadenas no están limitadas en tamaño, por lo que el único límite es la memoria de tu ordenador. Una cadena 
puede estar también vacía.
"""

s = ''

"""
Una situación que muchas veces se puede dar, es cuando queremos introducir una comilla, bien o sea simple o doble 
dentro de una cadena. Si lo hacemos de la siguiente forma tendremos error:

s = "Esto es una comilla doble " de ejemplo"

Para resolver este problemas debemos recurrir a las secuencias de escape. En python hay varias, la más importante 
es \", que nos permite incrustar comillas dentro de una cadena.
"""

s = "Esto es una comilla doble \" de ejemplo"
print(s) # -> Esto es una comilla doble " de ejemplo

"""
También podemos incluir un salto de línea dentro de una cadena, lo que significa que lo que esté después del salto 
se imprimirá en una nueva línea.
"""

s = "Primera línea\nSegunda línea"
print(s)
```

```python
# Formateo de cadenas

"""
Tal vez queramos declarar una cadena que contenga variables en su interior, como números o incluso otras cadenas, 
hay varias formas pero la que considero que es la mejor es usando las cadenas literales o f-strings. Permite 
incrustar expresiones dentro de cadenas.
"""

a = 5; b = 10

s = f"Los números son {a} y {b}. Su suma es: {a + b}"
print(s) # -> Los números son 5 y 10. Su suma es 15
```

```python
# Particularidades

## Indexación

"""
Las cadenas de texto están indexadas por posiciones desde 0 hasta n, así que si por ejemplo cogemos el ínidice 1 
estaríamos realmente cogiendo el segundo carácter de la cadena.

-> "Python"
    012345
    
Si usamos "Python"[1] cogemos y y así, hasta que literamente no haya otro carácter (generará un error si nos 
pasamos), y podemos empezar desde -1 si queremos hacerlo desde un orden inverso.
"""

s = "Python"

print(s[1]) # -> y
# print(s[8]) # -> al n

print(s[-2]) # -> o, lo estamos haciendo desde atras hacía adelante

## Longitud

"""
Ahora que sabemos la indexación de los strings, eso nos lleva a pensar cuántos caracteres tiene x cadena, esto 
nos lo permite la función len() es una función que nos permite saber la longitud en este caso de nuestra cadena, 
esto nos puede servir si depronto necesitamos saber la longitud de cierta cadena para no pasarnos y hacer la 
operación que necesitemos.

Más adelante se profundizará en las funciones y usos del len()
"""

print(len(s)) # -> 6

## Creación de cadenas a bases de otras

"""
Podemos crear cadenas más pequeñas a base de una más grande, indicando el primer elmento y el último que queremos 
tomar menos uno.
"""

x = "abcde"
print(x[0:2]) # -> ab

"""
Si no se indica ningún valor a la derecha de los : se llega hasta el final.
"""

x = "abcde"
print(x[2:]) # -> cde

"""
Es posible también crear subcadenas qué contengan elementos salteados y no contiguos añadiendo un tercer elemento 
entre []. Indica los elementos que se saltan. En el siguiente ejemplo se toman elementos del 0 al 5 de 2 en 2.
"""

x = "abcde"
print(x[0:len(x):2]) # -> ace, con la longitud de nuestra cadena

"""
Tampoco es necesario sabernos el tamaño de la cadena, y el segundo valor se podría omitir. El siguente ejemplo es 
igual al anterior.
"""

x = "abcde"
print(x[0::2]) # -> ace 
```
