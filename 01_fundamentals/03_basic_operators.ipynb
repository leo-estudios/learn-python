```python
# Operadores aritméticos

"""
Los operadores aritméticos son los más comunes que nos podemos encontrar, y nos permiten realizar operaciones 
aritméticas sencillas, como lo pueden ser la suma, resta o exponente.

Para el uso de estos operadores es tener en cuenta el orden de estos y cual se aplica primero, el orden de 
prioridad sería este:

- () -> Paréntesis
- ** -> Exponentes
- *, /, //, % -> Multiplicación, División, Cociente, Módulo
- +, - -> Suma, Resta
"""

## Suma
print(9 + 1) # -> 10

## Resta
print(4 - 8) # -> 4

## Multiplicación
print(10 * 2) # -> 20

## División
print(4 / 2) # -> 2

## Módulo
print(6 % 3) # -> 0

## Exponente
print(3 ** 3) # -> 27

## Cociente
print(6 // 2) # -> 3
```

```python
# Operadores de asignación

"""
Nos permiten realizar una operación y almacenar su resultado en la variable inicial. Podemos ver como realmente el 
único operador nuevo es el =. El resto son abreviaciones de otros operadores.
"""

## Asignación
x = 2
print(x)

## Asignación con suma
x += 3 # -> 5 --> x = x + 3
print(x)

## Asignación con resta
x -= 3 # -> 2 --> x = x - 3
print(x)

## Asignación con multiplicación
x *= 2 # -> 4 --> x = x * 2
print(x)

## Asignación con división
x /= 2 # -> 2 --> x = x / 2
print(x)

## Asignación con exponenciación
x **= 2 # -> 4 --> x = x ** 2
print(x)

## Asignación con cociente
x //= 2 # -> 2 --> x = x // 2
print(x)

## Asignación con módulo
x %= 2 # -> 0 --> x = x % 2
print(x)
```

```python
# Operadores relacionales

"""
Nos permiten saber la relación existente entre dos variables o valores. Se usan para saber si por ejemplo un número es 
mayor o menor que otro. Dado que estos operadores indican si se cumple o no una operación, el valor que devuelven es 
True o False.
"""

## Igual
print(3 == "3") # -> False

## Diferencia
print(4 != 9) # -> True

## Mayor qué
print(4 > 2) # -> True

## Menor qué
print(2 < 9) # -> True

## Mayor igual qué
print(3 >= 8) # -> False

## Menor igual qué
print(10 <= 9) # -> False
```

```python
# Operadores lógicos

"""
Nos permiten trabajar con valores tipo bool. Un valor booleano es un tipo que solo puede tomar valores True o False. 
Por lo tanto, estos operadores nos permiten realizar diferentes operaciones con estos tipos, y su resultado será 
otro booleano. Por ejemplo, 'True and True' usa el operador and, y su resultado será True.

- and -> Devuelve True si ambos elementos son True
- or -> Devuelve True si al menos un elemento es True
- not -> Devuelve el contrario, True si es False y viceversa
"""

## and
print(True and True) # -> True
print(True and False) # -> False
print(False and True) # -> False
print(False and False) # -> False

## or
print(True or True) # -> True
print(True or False) # -> True
print(False or True) # -> True
print(False or False) # -> False

## not
print(not True) # -> False
print(not False) # -> True
print(not not True) # -> True
```
